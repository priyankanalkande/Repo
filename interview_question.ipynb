```python
'''Q6. Rotate a list by k positions
Problem: Rotate the list to the right by k positions.
Input: A list and an integer k.
Output: The rotated list.
Sample Input
nums = [1, 2, 3, 4, 5], k = 2
Expected Output
[4, 5, 1, 2, 3]
Interview Hint: Watch for k larger than the list size.  '''

def rotate_list(nums, k):
    n = len(nums)
    k = k % n   # handle k larger than list size
    return nums[-k:] + nums[:-k]

# Sample Input
nums = [1, 2, 3, 4, 5]
k = 2
print(rotate_list(nums, k))   # Output: [4, 5, 1, 2, 3]
```

```python
'''Q7. Find all pairs with target sum
Problem: Return all unique pairs of numbers whose sum equals a target value.
Input: A list of integers and a target integer.
Output: A list of unique pairs.
Sample Input
nums = [2, 7, 4, 5, 3, 1, 6], target = 8
Expected Output
[(2, 6), (7, 1), (5, 3)]
Interview Hint: Interviewers care about avoiding duplicate pairs.  '''

def find_pairs(nums, target):
    seen = set()
    pairs = set()

    for num in nums:
        complement = target - num
        if complement in seen:
            # store pair in sorted order to avoid duplicates
            pairs.add(tuple(sorted((num, complement))))
        seen.add(num)

    return list(pairs)

# Sample Input
nums = [2, 7, 4, 5, 3, 1, 6]
target = 8
print(find_pairs(nums, target)) 
```

```python
'''Q9. Validate and convert numeric strings
Problem: Given values read from a CSV as strings, convert them into integers or floats where possible and mark
invalid values.
Input: A list like ["10", "5.5", "", "abc", "20"].
Output: A cleaned list and optionally a list of invalid values.
Sample Input
["10", "5.5", "", "abc", "20"]
Expected Output
[10, 5.5, None, None, 20]
Interview Hint: A good answer explains how blanks and bad values are handled '''

s=["10", "5.5", "", "abc", "20"]
result=[]
invalid=[]
for i in s:
    if i.strip()=="":
        result.append(None)
        invalid.append(i)
        continue
    try:
        result.append(int(i))
    except ValueError:
        try:
            # Try float if not integer
            result.append(float(i))
        except ValueError:
                result.append(None)
                invalid.append(i)

print(result)
print(invalid)
```

```python
'''Q10. Chunk a list into batches
Problem: Split a list into fixed-size batches for downstream processing.
Input: A list and batch size n.
Output: A list of lists.
Sample Input
nums = [1,2,3,4,5,6,7], n = 3
Expected Output
[[1,2,3],[4,5,6],[7]]
Interview Hint: Common in ETL batching and API pagination  '''

def chunk_list(nums, n):
    chunks = []
    for i in range(0, len(nums), n):
        chunks.append(nums[i:i+n])
    return chunks

# Sample Input
nums = [1, 2, 3, 4, 5, 6, 7]
n = 3
print(chunk_list(nums, n))   
```

```python
'''Q11. Write validate_record(record)
Problem: Create a function validate_record(record) that checks mandatory fields id, name, and email. id must be
numeric and email must contain @.
Input: A dictionary like {'id': '101', 'name': 'Anuj', 'email': 'anuj@test.com'}.
Output: Return True/False, or return a tuple such as (is_valid, errors).
Sample Input
{'id': '101', 'name': 'Anuj', 'email': 'anuj@test.com'}
Expected Output
(True, [])
Interview Hint: Returning detailed error messages is a strong answer. '''

def validate_record(record):
    errors = []

    # Check id
    if "id" not in record or not record["id"].isdigit():
        errors.append("Invalid or missing id")

    # Check name
    if "name" not in record or not record["name"].strip():
        errors.append("Invalid or missing name")

    # Check email
    if "email" not in record or "@" not in record["email"]:
        errors.append("Invalid or missing email")

    return (len(errors) == 0, errors)


# Sample Input
record1 = {'id': '101', 'name': 'Anuj', 'email': 'anuj@test.com'}
print(validate_record(record1))   # (True, [])

record2 = {'id': 'abc', 'name': '', 'email': 'anujtest.com'}
print(validate_record(record2))   # (False, ['Invalid or missing id', 'Invalid or missing name', 'Invalid or missing email'])
```

```python
'''Q14. Write a generator for file lines
Problem: Create a generator function that yields one line at a time from a file path.
Input: A file path string.
Output: A generator that yields stripped lines.
Sample Input
path = 'system.log'
Expected Output
yield each line one by one
Interview Hint: Interviewers like this because it is memory efficien  '''

def read_file_lines(path):
    """
    Generator function that yields one line at a time from a file.
    Strips newline characters and extra spaces.
    """
    try:
        with open(path, 'r') as f:
            for line in f:
                yield line.strip()
    except FileNotFoundError:
        print(f"File not found: {path}")
    except Exception as e:
        print(f"Error reading file: {e}")


# Sample Usage
path = 'system.log'

for line in read_file_lines(path):
    print(line)

'''🧩 Step-by-Step Explanation
Function Definition

def read_file_lines(path): → defines a generator function that takes a file path as input.

Open the File Safely

with open(path, 'r') as f: → opens the file in read mode.

The with statement ensures the file is closed automatically after use.

Iterate Line by Line

for line in f: → loops through the file one line at a time.

This is memory‑efficient because it doesn’t load the entire file into memory.

Yield Each Line

yield line.strip() → returns one line at a time, removing leading/trailing spaces and newline characters.

Because it uses yield, the function becomes a generator — you can loop over it without storing all lines at once.

Error Handling

except FileNotFoundError: → prints a message if the file doesn’t exist.

except Exception as e: → catches any other errors (like permission issues) and prints them.

Usage Example

for line in read_file_lines(path): → loops through the generator, printing each line.

This way, you process the file line by line.'''
```

```python
'''Q15. Retry wrapper
Problem: Write a reusable function that retries another function up to 3 times if it raises an exception.
Input: A callable and retry count.
Output: The function result or a final failure after retries.
Sample Input
retry(read_file, 3)
Expected Output
Either the successful result or final raised exception
Interview Hint: Good answers usually use try/except inside a loop. '''

def retry(func, retries=3, *args, **kwargs):
    """
    Retry wrapper: runs a function up to 'retries' times.
    If it succeeds, return the result.
    If it fails every time, raise the last exception.
    """
    last_exception = None
    for attempt in range(1, retries + 1):
        try:
            # Try calling the function
            return func(*args, **kwargs)
        except Exception as e:
            print(f"Attempt {attempt} failed: {e}")
            last_exception = e
    # After all retries, raise the last exception
    raise last_exception


# Example function to test
def read_file(path):
    with open(path, 'r') as f:
        return f.read()


# Usage
try:
    result = retry(read_file, 3, "system.log")
    print("File content loaded successfully")
except Exception as e:
    print("Final failure after retries:", e)

'''🧩 Explanation
Function definition

retry(func, retries=3, *args, **kwargs) → accepts any function, retry count, and its arguments.

Loop through attempts

for attempt in range(1, retries+1) → tries up to the given number of times.

Try/except block

If the function succeeds, return the result immediately.

If it fails, catch the exception, print the error, and retry.

Final failure

If all retries fail, raise the last exception so the caller knows it didn’t succeed.'''
```

```python
'''Q16. Safe dictionary getter
Problem: Write a helper function safe_get(data, path, default=None) to read nested dictionary keys safely.
Input: Nested dictionary and a path like ['order', 'customer', 'id'].
Output: The nested value if present, otherwise default.
Sample Input
data = {'order': {'customer': {'id': 101}}}
Expected Output
101
Interview Hint: This is practical for nested JSON APIs. '''

def safe_get(data, path, default=None):
    """
    Safely get a nested value from a dictionary.
    data: dictionary
    path: list of keys (e.g. ['order','customer','id'])
    default: value to return if any key is missing
    """
    current = data
    for key in path:
        if isinstance(current, dict) and key in current:
            current = current[key]
        else:
            return default
    return current

# Sample Input
data = {'order': {'customer': {'id': 101}}}
print(safe_get(data, ['order', 'customer', 'id']))       # Output: 101
print(safe_get(data, ['order', 'customer', 'name']))     # Output: None
print(safe_get(data, ['order', 'customer', 'name'], 'N/A'))  # Output: 'N/A'

'''🧩 Explanation
Iterate through path keys

Start with the dictionary (current = data).

For each key in the path, check if it exists in the current dictionary.

Check type

Ensure current is still a dictionary before accessing keys.

Return default if missing

If a key is missing, return the default value immediately.

Return final value

If all keys are found, return the nested value.'''
```

```python
'''Q17. Decorator to log execution time
Problem: Write a decorator that prints or logs the execution time of a function.
Input: Any function.
Output: Wrapped function output plus timing information.
Sample Input
@timer
def load_data(): ...
Expected Output
Execution time: 0.021 sec
Interview Hint: A simple time module solution is enough. '''

import time
from functools import wraps

def timer(func):
    @wraps(func)   # preserves original function name and docstring
    def wrapper(*args, **kwargs):
        start = time.time()              # record start time
        result = func(*args, **kwargs)   # run the actual function
        end = time.time()                # record end time
        print(f"Execution time: {end - start:.3f} sec")
        return result
    return wrapper


# Sample Usage
@timer
def load_data():
    # Simulate some work
    time.sleep(0.02)
    return "Data loaded"

print(load_data())

'''🧩 Explanation
Import time

time.time() gives the current time in seconds.

Decorator definition

def timer(func): → defines the decorator.

Inside, wrapper is the function that will replace the original.

Measure execution time

Record start time before calling the function.

Record end time after it finishes.

Print the difference.

Return result

The wrapper returns the original function’s result so behavior is unchanged.

@wraps(func)

Keeps the original function’s name and docstring (good practice in interviews).'''
```

```python
'''Q18. Normalize text utility
Problem: Write a function normalize_name(name) that trims spaces, removes repeated internal spaces, and converts
text to title case.
Input: A raw string like " anuj SHAHDEO ".
Output: A cleaned string.
Sample Input
" anuj SHAHDEO "
Expected Output
"Anuj Shahdeo"
Interview Hint: Common cleansing task in data pipelines. '''

def normalize_name(name):
    """
    Normalize a name string:
    - Trim leading/trailing spaces
    - Remove repeated internal spaces
    - Convert to title case
    """
    # Step 1: strip leading/trailing spaces
    cleaned = name.strip()

    # Step 2: split into words and rejoin with single space
    cleaned = " ".join(cleaned.split())

    # Step 3: convert to title case
    return cleaned.title()


# Sample Input
print(normalize_name(" anuj SHAHDEO "))   # Output: "Anuj Shahdeo"
print(normalize_name("   priya   kumar   "))  # Output: "Priya Kumar"
```

```python
'''Q19. Count rows with missing mandatory columns
Problem: Given CSV rows, count how many rows are missing email or phone.
Input: CSV file with headers id,name,email,phone.
Output: Total missing count and optionally the row numbers.
Sample Input
id,name,email,phone
1,A,aa@test.com,9999
2,B,,8888
3,C,cc@test.com,
4,D,dd@test.com,7777
Expected Output
missing_count = 2
Interview Hint: A good solution uses csv.DictReader '''

import csv

def count_missing(path):
    """
    Count rows in a CSV file that are missing email or phone.
    Returns (missing_count, row_numbers).
    """
    missing_count = 0
    missing_rows = []

    with open(path, newline='') as f:
        reader = csv.DictReader(f)
        for i, row in enumerate(reader, start=2):  # start=2 because row 1 is header
            if not row.get("email") or not row.get("phone"):
                missing_count += 1
                missing_rows.append(i)

    return missing_count, missing_rows


# Sample Usage
path = "sample.csv"
missing_count, missing_rows = count_missing(path)
print("Missing count:", missing_count)
print("Missing rows:", missing_rows)
```

```python
'''Q20. CSV to dictionary records
Problem: Read a CSV file and convert each row into a dictionary using the header row.
Input: A CSV file path.
Output: A list of dictionaries.
Sample Input
id,name,dept
1,A,IT
2,B,HR
ExpectedOutput
[{'id':'1','name':'A','dept':'IT'},{'id':'2','name':'B','dept':'HR'}]
InterviewHint:Thisisbasicbutveryimportant  '''

import csv

def csv_to_dicts(path):
    """
    Read a CSV file and convert each row into a dictionary.
    Returns a list of dictionaries.
    """
    records = []
    with open(path, newline='') as f:
        reader = csv.DictReader(f)   # uses header row as keys
        for row in reader:
            records.append(row)
    return records


# Sample Usage
path = "sample.csv"
print(csv_to_dicts(path))

'''🧩 Explanation
csv.DictReader

Reads the CSV file line by line.

Uses the header row (id,name,dept) as dictionary keys.

Loop through rows

Each row is already a dictionary like {'id':'1','name':'A','dept':'IT'}.

Append it to the list.

Return the list

Final result is a list of dictionaries, one per row.'''
```

```python
'''Q21.Removeduplicaterowsbycustomer_id
Problem:FromaCSVfile,keeponlythefirstrecordforeachcustomer_id.
Input:CSVrowswithpossibleduplicatecustomer_idvalues.
Output:Adeduplicatedlistoroutputfile.
SampleInput
customer_id,name
101,A
102,B
101,A2
ExpectedOutput
customer_id,name
101,A
102,B
InterviewHint:Set-basedtrackingisexpected '''

import csv

def deduplicate_by_customer(path):
    """
    Read a CSV file and keep only the first record for each customer_id.
    Returns a list of unique rows.
    """
    seen = set()
    unique_rows = []

    with open(path, newline='') as f:
        reader = csv.DictReader(f)
        for row in reader:
            cid = row.get("customer_id")
            if cid not in seen:
                seen.add(cid)
                unique_rows.append(row)

    return unique_rows


# Sample Usage
path = "sample.csv"
result = deduplicate_by_customer(path)
print(result)

'''🧩 Explanation
Use csv.DictReader

Reads each row as a dictionary with keys from the header (customer_id, name).

Track seen IDs

Use a set() called seen to remember which customer_id values have already appeared.

Keep only first occurrence

If customer_id not in seen, add it and keep the row.

If already in seen, skip (removes duplicates).

Return unique rows

Final result is a list of dictionaries with no duplicate customer_id.'''
```

```python
'''Q22.Totalamountpercustomer
Problem:ReadtransactionCSVdataandcalculatetotalamountpercustomer.
Input:CSVwithcustomer_idandamountcolumns.
Output:Adictionaryoroutputfilewithtotalamountpercustomer.
SampleInput
customer_id,amount
101,100
102,50
101,75
ExpectedOutput
{'101':175.0, '102':50.0}
InterviewHint:Candidateshouldconvertamounttofloatsafely '''

import csv

def total_amounts(path):
    """
    Read transaction CSV data and calculate total amount per customer.
    Returns a dictionary {customer_id: total_amount}.
    """
    totals = {}
    with open(path, newline='') as f:
        reader = csv.DictReader(f)
        for row in reader:
            cid = row.get("customer_id")
            try:
                amt = float(row.get("amount", 0))   # safely convert to float
            except ValueError:
                amt = 0.0   # handle invalid numbers gracefully

            totals[cid] = totals.get(cid, 0.0) + amt

    return totals


# Sample Usage
path = "transactions.csv"
print(total_amounts(path))

'''🧩 Explanation
Use csv.DictReader

Reads each row as a dictionary with keys from the header (customer_id, amount).

Initialize totals dictionary

Keys = customer IDs, values = accumulated amounts.

Convert amount safely

float(row.get("amount", 0)) → ensures numeric conversion.

If conversion fails (bad data), default to 0.0.

Accumulate totals

totals[cid] = totals.get(cid, 0.0) + amt → adds the amount to the existing total.

Return dictionary

Final result is {customer_id: total_amount}.'''
```

```python
'''Q23.Splitvalidandinvalidrows
Problem:ReadaCSVfile,validaterows,andwritevalidrowsandinvalidrowstoseparateoutputfiles.
Input:CSVrowscontainingid,email,amount.
Output:Twofiles:valid.csvandinvalid.csv,ortwoin-memorylists.
SampleInput
id,email,amount
1,a@test.com,100
x,b@test.com,90
3,invalid,20
ExpectedOutput
valid_rows = 1,invalid_rows = 2
InterviewHint:ThisisaveryrealisticETLinterviewquestion '''

import csv

def split_valid_invalid(path):
    valid_rows = []
    invalid_rows = []

    with open(path, newline='') as f:
        reader = csv.DictReader(f)
        for row in reader:
            # Validation: id numeric, email contains '@', amount numeric
            if row["id"].isdigit() and "@" in row["email"] and row["amount"].replace('.', '', 1).isdigit():
                valid_rows.append(row)
            else:
                invalid_rows.append(row)

    return valid_rows, invalid_rows


# Sample Usage
path = "sample.csv"
valid, invalid = split_valid_invalid(path)
print("Valid rows =", len(valid))
print("Invalid rows =", len(invalid))

''' 🧩 Explanation
Read CSV with DictReader → each row is a dictionary.

Validation checks:

id.isdigit() → ensures id is numeric.

"@" in email → ensures email looks valid.

amount.replace('.', '', 1).isdigit() → allows numbers with decimals.

Split rows → if all checks pass, add to valid_rows; otherwise, add to invalid_rows.

Return results → gives two lists, and you can count them easily. '''
```

```python
'''Q24.Highestsalaryrecord
Problem:ReadasalaryCSVfileandreturnthefullrecordwiththehighestsalary.
Input:CSVwithemployee_id,employee_name,salary.
Output:Thefullrecordfortheemployeewiththehighestsalary.
SampleInput
employee_id,employee_name,salary
1,A,5000
2,B,7000
3,C,6500
ExpectedOutput
{'employee_id':'2','employee_name':'B','salary':'7000'}
InterviewHint:Manyinterviewerswantcsv.DictReaderplusmaxlogic '''

import csv

def highest_salary(path):
    """
    Read a salary CSV file and return the full record
    with the highest salary.
    """
    with open(path, newline='') as f:
        reader = csv.DictReader(f)
        # Use max() with key = salary converted to float
        highest = max(reader, key=lambda row: float(row["salary"]))
    return highest


# Sample Usage
path = "salary.csv"
print(highest_salary(path))

''' 🧩 Explanation
Use csv.DictReader

Reads each row as a dictionary ({'employee_id':'1','employee_name':'A','salary':'5000'}).

Find maximum by salary

max(reader, key=lambda row: float(row["salary"])) → finds the row with the largest salary.

Convert salary to float to ensure numeric comparison.

Return the record

The result is the full dictionary for the employee with the highest salary.'''
```

```python
'''Q25.Readpipe-delimitedfile
Problem:Processapipe-delimitedfileinsteadofcomma-separatedinputandcountblankmandatoryfields.
Input:Adelimitedtextfilelikeid|name|dept.
Output:Blankfieldcountorrejectedrows.
SampleInput
id|name|dept
1|A|IT
2||HR
3|C|
ExpectedOutput
blank_count=2
InterviewHint:Candidateshouldknowthedelimiterargumentincsvmodule. '''

import csv

def count_blank_fields(path):
    blank_count = 0
    with open(path, newline='') as f:
        # Specify delimiter as '|'
        reader = csv.DictReader(f, delimiter='|')
        for row in reader:
            # Check each mandatory field
            for value in row.values():
                if value.strip() == "":
                    blank_count += 1
    return blank_count


# Sample Usage
path = "sample.txt"
print("Blank count =", count_blank_fields(path))

'''🧩 Explanation
delimiter='|'

Tells csv.DictReader to split fields using | instead of ,.

Loop through rows

Each row is a dictionary like {'id':'1','name':'A','dept':'IT'}.

Check for blanks

If any field is empty ("" after stripping spaces), increase the counter.

Return total

Gives the total number of blank mandatory fields across all rows.'''
```
