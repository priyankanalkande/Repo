```python
'''Q57. Cheapest Product per Category

Use products.txt

Task:

Find lowest price product in each category '''

def is_number(val):
    try:
        return int(val)
    except ValueError:
        return None

cheapest = {}

with open("products.txt", "r") as f:
    for line in f:
        parts = line.strip().split(",")
        if len(parts) != 4:
            continue
        id, name, price, category = parts

        int_price = is_number(price)
        if int_price is None:
            continue  # skip invalid prices

        # If category not seen before OR current product is cheaper
        if category not in cheapest or int_price < cheapest[category][1]:
            cheapest[category] = (name, int_price)

# Print results
for cat, (prod_name, prod_price) in cheapest.items():
    print(f"{cat}: {prod_name} ({prod_price})")
```

```python
'''Section 7: Real Pipeline Simulation

Q64. Mini ETL Pipeline (VERY IMPORTANT)

Use employees.txt

Steps:
Read file
Clean data
Transform (add bonus)
Store in list of dict
Write to output file
Log errors '''

def num(val):
    try:
         return int(val)
    except ValueError:
        return None
    
emp=[]
errors=[]

with open("employee.txt","r") as f,open("emp64.txt","w") as f1,open("error64.log","w") as f2:
    for line in f:
        lines=line.strip().split(",")
        
        if len(lines)!=3:
            errors.append(f"Invalid format :{lines}")
            continue
        id,name,sal =lines

        int_sal=num(sal)
        if int_sal is None:
            errors.append(f"Invalid salary for {id}:{sal}")
            continue

        bonus=int_sal*0.10
        total=int_sal+bonus

        emp_dict ={
            "id":id,
            "Name":name,
            "salary":int_sal,
            "Bonus":bonus,
            "total":total
        }

        emp.append(emp_dict)
    for e in emp:
         f1.write(str(e)+"\n")
    for e1 in errors:

         f2.write(str(errors)+"\n")
print(emp)
print(errors)
```
