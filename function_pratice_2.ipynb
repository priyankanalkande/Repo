```python
'''Problem 3: Consecutive Drop Detection

You are given a list of numbers representing daily sales.

Create a function that:
	•	Loops through the list
	•	Tracks consecutive drops
	•	Returns "Critical" if sales drop 3 times in a row
	•	Else returns "Stable"

Focus: counter reset logic'''

def num(number):
    count=0
    prev=number[0]
    for curr in number[1:]:
        if curr< prev:
            count+=1
            if count==3:
                 return "Critical"
        else:
            count=0
        prev=curr
    return "stable"

lst=[100,90,80,70,60]
num(lst)
```

```python
'''Problem 4: Access Escalation Logic

A list represents login attempts:
	•	"user" → normal user
	•	"admin" → admin user
	•	"blocked" → account blocked

Write a function that:
	•	Loops through attempts
	•	If "blocked" appears → return "Account Locked"
	•	If "admin" appears after a "user" → return "Privilege Escalation"
	•	Else return "Normal Access"

Focus: order matters'''

def access(users):
    user_seen=0
    for i in users:
        if i =="blocked":
            return "Account locked"
        if i=="user":
            user_seen=1
        if i=="admin" and user_seen:
                return "Privilege Escalation"
    return "Normal Access"

lst=["admin","user","blocked"]
access(lst)
```

```python
'''Problem 5: First Repeating Pattern

Given a list of values, write a function that:
	•	Returns the first value that repeats
	•	Must stop scanning as soon as repetition is detected
	•	If no repetition exists, return "Unique"'''

def first_repeating(values):
    seen = set()                     # keep track of values we’ve seen
    for v in values:                 # loop through each value
        if v in seen:                # repetition detected
            return v                 # return immediately
        seen.add(v)                  # otherwise record it
    return "Unique"                  # if no repetition found

# Example usage
lst1 = [10, 20, 30, 20, 40]
lst2 = [1, 2, 3, 4, 5]

print(first_repeating(lst1))   # 20
print(first_repeating(lst2))   # Unique
```

```python
'''Problem 7: Write a function that:
	•	Takes initial resource count and a list of requests
	•	Loops through requests
	•	Deducts request if possible
	•	Returns "Exhausted" immediately if resources go below 0
	•	Else return remaining resources

Focus: simulation + early failure'''

def resource_exhaustion(resources, requests):
    #resources = initial
    for req in requests:          # loop through each request
        resources -= req          # deduct from resources
        if resources < 0:         # check if exhausted
            return "Exhausted"    # early return
    return resources              # return remaining if all requests succeed

lst1=[5, 8, 6,2]
lst2=20
print(resource_exhaustion(lst2,lst1))   # Exhausted

```

```python
'''Problem 8: Log Sequence Validator
Logs are expected in this order:
	1.	"START"
	2.	"PROCESS"
	3.	"END"

Create a function that:
	•	Loops through log entries
	•	Returns "Invalid Sequence" if order is violated
	•	Returns "Valid Sequence" if correct

Focus: state machine logic'''

def seq_validator(log):
    expected=["start","process","end"]
    state =0
    for i in log:
        if state>=len(expected):
            return "Invalid Sequence" 
        if i ==expected[state]:
            state+=1
        else: 
              return "Invalid Sequence"
    if state ==len(expected):
            return "valid Sequence"
    else:
         return "Invalid Sequence"
    
lst =["start","process","end"]
seq_validator(lst)
```

```python
'''Problem 9: Suspicious Activity Detection
You are given a list of actions:
	•	"login"
	•	"view"
	•	"delete"

Write a function that:
	•	Loops through actions
	•	Returns "Suspicious" if "delete" happens before "login"
	•	Else return "Normal" '''

def action(activity):
    actions=0
    for i in activity:
        if i=="delete":
            actions=1
        
        if i=="login" and actions:
            return "Suspicious"
    return "normal"


lst=["delete","login","view"]
action(lst)
```

```python
''' Problem 11: Fail-Fast Data Validation
Given a list of records (each record is a dictionary):
Create a function that:
	•	Loops through records
	•	Returns "Invalid Record" if any record:
	•	Missing required key "id" OR
	•	"id" is None
	•	Else return "All Valid"

Focus: fail-fast return'''

def validate_records(records):
    for record in records:                  # loop through each record
        if "id" not in record:              # missing key
            return "Invalid Record"
        if record["id"] is None:            # id is None
            return "Invalid Record"
    return "All Valid"                      # if loop finishes safely

# Example usage
records1 = [{"id": 1}, {"id": 2}, {"id": 3}]          # All Valid
records2 = [{"id": 1}, {"name": "test"}, {"id": 3}]   # Invalid Record (missing id)
records3 = [{"id": None}, {"id": 2}]                  # Invalid Record (id is None)

print(validate_records(records1))
print(validate_records(records2))
print(validate_records(records3))
```

```python
''' Problem 12: Threshold Breach
A sensor sends readings.

Write a function that:
	•	Loops through readings
	•	Tracks how many times reading exceeds 100
	•	Returns "Shutdown" if breaches occur more than twice
	•	Else return "Operational"

Focus: count + early exit'''

def num(number):
    breaches=0
    for i in number:
        if i >100:
            breaches+=1
            if breaches>2:
                return "shutdown"
    return "Operational"

lst=[80,200,30,20,200,200]
num(lst)

```
