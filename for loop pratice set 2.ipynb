```python
'''5. Rule:
    If Any values is 0 -> System Fail 
    If All values >= 50 -> System Ok 
    Else -> system WARNING 

    values=[70,55,0,80]'''

values=[69,40,70,0,40]
fail=False
all_ok=True
for v in values:
    if v == 0:
        fail=True
    if v < 50:
        all_ok=False

if fail:
    print("System Failed")
elif all_ok:
    print("System ok")
else:
    print("SYSTEM WARNING")
```

```python
'''6. scores=[65,40,70,55,80,59,10]

Rule: 
- Values >= 60 -> Positive vote 
- If postive votes > negative : PASS 
- Fail '''

scores=[65,40,70,55,80,59,10]
positive = 0
negative=0
for v in scores:
    if v>=60:
        positive +=1
    else:
        negative +=1
if positive > negative:
    print("pass")
else:
    print("fail")

```

```python
'''9. Resouce limitation 
    rule:
    - Total resouces = 100 
    - consume values one by one 
    - If resource goes negative -> STOP 

    requests= [20,30,40,25]
    resouces = 100 '''

requests = [20, 30, 40,50]
resources = 100

for req in requests:
    resources -= req
    print(f"{req}, Remaining = {resources}")
    if resources < 0:
        print("STOP")
        break

```

```python
'''
12. Proformanace Review 
    ratings =[4,3,5,2,4]
    Rules :
        - if rating < 3
            - If more than one rating < 3 -> NO BONUS 
            - ELSE -> REVIEW 
        - ELSE 
            if rating  >= 4 --> BONUS 
            else -> NO BONUS '''

ratings = [4, 3, 5, 2, 4]

low_count = 0     # count of ratings < 3
bonus_flag = 0    # 0 = no bonus, 1 = bonus candidate

for r in ratings:
    if r < 3:
        low_count += 1
    elif r >= 4:
        bonus_flag = 1   # mark that at least one bonus rating exists

# Final decision
if low_count > 1:
    print("NO BONUS")
elif low_count == 1:
    print("REVIEW")
else:
    if bonus_flag == 1:
        print("BONUS")
    else:
        print("NO BONUS")

```

```python
'''13. levels = [100,90,80,85,70]
    Rules :
        - If values drop consecutively:
            - if 3 continious drop -> Critical 
            - Else - Warning 
        - Else -> NORMAL '''

levels = [100,90,80,85,70]
drop_count = 0
max_drop=0
prev=levels[0]

for curr in levels[1:]:
    if(curr<prev):
        drop_count+=1
        if(drop_count>max_drop):
            max_drop = drop_count
    else:
        drop_count =0
    prev =curr
if max_drop >=3:
    print("critical")
elif max_drop > 0:
    print("Warning") 
else: 
    print("Normal")
```

```python
'''# Recursion Questions 

1. Print Number N-1 
2. Print Number 1 - N 
3. Sum of First N numbers 
4. Factorial 
5. Count Digits 
6. Sum of Digits = Input 1234 , Output : 10 
7. Power of (x^n) = Input x=2, n=3 , Output : 8 
8. Palindrom no - True or False '''
```

```python
#1. Print Number N-1 

def number(n):
    if n==0:
        return 
    print(n)
    return number(n-1)

number(5)
```

```python
#2. Print Number 1 - N 

def num(n):
    if n==0:
        return
    num(n-1)
    print(n)

num(5)
```

```python
#3. Sum of First N numbers 

def total(n):
    if n==0:
        return 0
    return n + total(n-1)

total(5)
```

```python
#4. Factorial 
#5!=5!*4!*3!*2!*1! =120

def fact(n):
    if n==0 or n==1:
        return 1
    return n* fact(n-1)

fact(5)
     
```

```python
#5. Count Digits 

def count_digits(n):
    if n==0:
        return 0
    return 1+count_digits(n//10)
print(count_digits(12345))
```

```python
'''6. Sum of Digits = Input 1234 , Output : 10 
'''

def sum_of_digits(n):
    if n == 0:              # base case
        return 0
    return (n % 10) + sum_of_digits(n // 10)

sum_of_digits(1234)
```

```python
'''7. Power of (x^n) = Input x=2, n=3 , Output : 8 '''
def power(x,n):
    if n==0:
        return 1
    return x *power(x,n-1)

power(2, 3)
```

```python
'''8. Palindrom no - True or False '''

num = int(input("Enter a number: "))
original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if original == reverse:
    print("Palindrome Number")
else:
    print("Not Palindrome")
```
