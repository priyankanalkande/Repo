```python
i = 1
while i < 106:
    if i % 3 == 0:
        print(f"{i} FIZZ")

        if i % 5 == 0:
            print(f"{i} BUZZ")
        else:
            print(f"{i} not BUZZ")

        if i % 7 == 0:
            print(f"{i} JAZZ")
        else:
            print(f"{i} not JAZZ")

        if i % 3 == 0 and i % 5 == 0 and i % 7 == 0:
            print(f"{i} FIZZBUZZJAZZ")

    else:
        print(f"{i} no is not divide by 3")

    i += 1   # important: increment to avoid infinite loop
```
