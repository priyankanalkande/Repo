```python
s={
    101:{"name":"Priyanka","maths":85,"science":90,"english":75},
    102:{"name":"Nilam","maths":88,"science":60,"english":85},
    103:{"name":"Priya","maths":30,"science":0,"english":5}
}

#1. Print all student names 

for key in s.values():
    print(key['name'])

#2. Print all student Ids 

print(list(s.keys()))

#3. Print Full student details 
for k,v in s.items():
    print(k,v)

#4. Add new Student 
s.update({104:{"name":"Priya","maths":30,"science":0,"english":5}})
print(s)

#5. Update marks of a student  "Priyanka" maths= 100
s.update({101:{"name":"Priyanka","maths":100,"science":90,"english":75}}) 
print(s)

s[102]['maths']=100
print(s)


#6. Calculate total marks of each student
for student_id, details in s.items():
    total = details["maths"] + details["science"] + details["english"]
    print(f"ID: {student_id}, Name: {details['name']}, Total Marks: {total}")


#7. Find the topper of the class
topper_name = ""
topper_id = 0
highest_marks = 0

for student_id, details in s.items():
    total = details["maths"] + details["science"] + details["english"]
    if total > highest_marks:
        highest_marks = total
        topper_name = details["name"]
        topper_id = student_id

print(f"Topper: {topper_name} (ID: {topper_id}), Total Marks: {highest_marks}")

s = {
    101: {"name": "Priyanka", "maths": 85, "science": 90, "english": 75},
    102: {"name": "Nilam", "maths": 88, "science": 60, "english": 85},
    103: {"name": "Priya", "maths": 30, "science": 0, "english": 5},
    104: {"name": "Priya", "maths": 30, "science": 0, "english": 5}
}

#8. Remove a student who left the institute (say ID 103)
del s[103]
print("After removal:", s)

#9. Check if a student exists (say ID 102)
if 102 in s:
    print("Student exists:", s[102]["name"])
else:
    print("Student does not exist")

#10. Sort students by total marks

# Make a list of (id, name, total marks)
students = []
for student_id, details in s.items():
    total = details["maths"] + details["science"] + details["english"]
    students.append((student_id, details["name"], total))

# Sort by total marks (highest first)
students.sort(key=lambda x: x[2], reverse=True)

# Print results
for sid, name, total in students:
    print(f"ID: {sid}, Name: {name}, Total Marks: {total}")


```
