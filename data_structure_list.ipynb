```python
'''56.	Sessionization (simplified)
Input: list of event times (minutes) sorted
Rule: new session if gap > 30
Output: list of sessions as list of lists.'''

events = [5, 10, 20, 60, 65, 120]   # sorted list of event times
sessions = []
current = []

for time in events:
    if not current:                 # start first session
        current.append(time)
    elif time - current[-1] > 30:   # gap > 30 → new session
        sessions.append(current)
        current = [time]
    else:                           # same session
        current.append(time)

# add last session
if current:
    sessions.append(current)

print(sessions)
```

```python
'''Question 1 — Data Cleaning Pipeline 

You receive user ages from API:

ages = ["25", "30", "", "18", None, "40"]

Task
	1.	Remove empty and None values using filter().
	2.	Convert remaining to integers using map().
	3.	Add new age 50 using append().
	4.	Sort ages.
	5.	Print:
	    •	total users (len)
	    •	max age
	    •	min age
	    •	sum of ages'''

ages = ["25", "30", "", "18", None, "40"]     

result=[]
for a in ages:
    if a!="" and a!=None:
        result.append(int(a))
        
result.append(50)
result.sort()

#sum_0f_ages=sum(result)
print(result)
print("length",len(result))
print(max(result))
print(min(result))
print(result)
print(sum(result))     
```

```python
ages = ["25", "30", "", "18", None, "40"]

# 1. Remove empty and None values using filter()
filtered = filter(lambda x: x not in ("", None), ages)

# 2. Convert remaining to integers using map()
result = list(map(int, filtered))

# 3. Add new age 50 using append()
result.append(50)

# 4. Sort ages
result.sort()

# 5. Print required stats
print("Total users:", len(result))
print("Max age:", max(result))
print("Min age:", min(result))
print("Sum of ages:", sum(result))
print("Sorted ages:", result)
```

```python
'''Question 2 — E-commerce Order Processing 

orders = [1000, 2500, 500, 1200, 800]

Task
	1.	Add 18% tax using map().
	2.	Remove orders below 1000 using filter().
	3.	Insert priority order 5000 at index 0.
	4.	Find index of highest order.
	5.	Reverse list using reversed() (without modifying original).'''

orders = [1000, 2500, 500, 1200, 800]

# 1. Add 18% tax using map()
orders_with_tax = list(map(lambda x: int(x * 1.18), orders))

# 2. Remove orders below 1000 using filter()
filtered_orders = list(filter(lambda x: x >= 1000, orders_with_tax))

# 3. Insert priority order 5000 at index 0
filtered_orders.insert(0, 5000)

# 4. Find index of highest order
highest_index = filtered_orders.index(max(filtered_orders))

# 5. Reverse list using reversed() (without modifying original)
reversed_orders = list(reversed(filtered_orders))

# Print results
print("Orders after tax and filtering:", filtered_orders)
print("Index of highest order:", highest_index)
print("Reversed list:", reversed_orders)
print(orders_with_tax)
```

```python
'''Question 3 — Log Analysis
logs = ["INFO", "ERROR", "INFO", "WARN", "ERROR", "DEBUG"]

Task
	1.	Count number of "ERROR" using count().
	2.	Convert all logs to uppercase using map().
	3.	Check:
	    •	if any ERROR exists (any)
	    •	if all logs are INFO (all)
	4.	Remove first DEBUG using remove().'''

logs = ["INFO", "ERROR", "INFO", "WARN", "ERROR", "DEBUG"]

#step1:-Count number of "ERROR" using count().

count=logs.count("ERROR")

#step 2:-2.	Convert all logs to uppercase using map().

uppers=list(map(lambda x: x.upper(),logs))

'''step 3.	Check:
	    •	if any ERROR exists (any)
	    •	if all logs are INFO (all)'''

any_error=any(u=="ERROR" for u in uppers)
all_info=all(u=="INFO" for u in uppers)

#4.	Remove first DEBUG using remove().
uppers.remove("DEBUG")

print(count)
print(uppers)
print(any_error)
print(all_info)

```

```python
'''Question 4 — Student Ranking System 

students = ["Anuj", "Amit", "Riya"]
marks    = [90, 80, 95]

Task
	1.	Combine using zip().
	2.	Sort by marks using sorted().
	3.	Add rank numbers using enumerate(start=1).
	4.	Convert result into list.
'''
students = ["Anuj", "Amit", "Riya"]
marks    = [90, 80, 95]

#2.	Sort by marks using sorted().
s=sorted(marks,reverse=1)

#	1.	Combine using zip().
z=list(zip(students,s))

#3.	Add rank numbers using enumerate(start=1).

result=list(map(lambda i: (i[0],i[1]), enumerate(z,start=1)))

print(z)
print(result)
```

```python
'''Question 5 — ETL File Path Generator 


files = ["data1.csv", "data2.csv", "data3.csv"]

Task
	1.	Generate S3 paths using map().
	2.	Copy original list using copy().
	3.	Extend list with backup files.
	4.	Reverse final list.'''

files = ["data1.csv", "data2.csv", "data3.csv"]

#1.	Generate S3 paths using map().
s3_files = list(map(lambda x: "s3://my-bucket/" + x, files))

#2.	Copy original list using copy().
copy_files=files.copy()

#3.	Extend list with backup files.
backup_files=[]
backup_files.extend(s3_files)
backup_files.extend(copy_files)

#4.	Reverse final list.
backup_files.reverse()

print(backup_files)
print(s3_files)
print(copy_files)
```

```python
'''Question 8 — API Response Flattening 

data = [
    {"id":1,"name":"A"},
    {"id":2,"name":"B"},
    {"id":3,"name":"C"}
]

Task
	1.	Extract all ids using map().
	2.	Extract all names.
	3.	Combine ids and names using zip().
	4.	Convert to list.'''

data = [
    {"id":1,"name":"A"},
    {"id":2,"name":"B"},
    {"id":3,"name":"C"}
]

#1.	Extract all ids using map().
id=list(map(lambda x: x["id"],data))

#2.	Extract all names.
name=list(map(lambda x: x["name"],data))

#3.	Combine ids and names using zip().
z=list(zip(id,name))

print(z)
print(id)
print(name)
```

```python
'''Question 9 — Real-Time Dashboard Validation 

values = [100, 200, -10, 300, 0]

Task
	1.	Filter invalid values (<=0).
	2.	Check using:
	    •	any() if negative exists
	    •	all() if valid
	3.	Add fallback value using insert().'''

values = [100, 200, -10, 300, 0]

#1.	Filter invalid values (<=0).
valid=list(filter(lambda x: x>=0,values))

'''	2.	Check using:
	    •	any() if negative exists
	    •	all() if valid'''

nigative= any(i<=0  for i in values)
all_valid= all(i>0 for i in values)

#3.	Add fallback value using insert().
valid.insert(0,99)

print(valid)
print(nigative)
print(all_valid)
```

```python
'''Question 10 — Complex List Transformation 

raw = ["  anuj ", " AMIT", "riya  ", ""]

Task
	1.	Remove empty values using filter().
	2.	Strip spaces using map(str.strip).
	3.	Capitalize names.
	4.	Enumerate names starting from 1.
	5.	Convert result into list of tuples.'''

raw = ["  anuj ", " AMIT", "riya  ", ""]

#1.	Remove empty values using filter().
val=list(filter(lambda x: x!="",raw))

#2.	Strip spaces using map(str.strip).
valid=list(map (str.strip,val))

#3.	Capitalize names.
cap=list(map(str.title,valid))

#4.	Enumerate names starting from 1.
names=list(map(lambda x:(x[0],x[1]), enumerate(cap,start=1)))

print(val)
print(valid)
print(cap)
print(names)
```

```python
'''Question 11:

ids   = [101,102,103,104]
names = ["Amit","Riya","Gaurav"]
city  = ["kolkata","Delhi","Pune","Goa","Mumbai"]

Task
	1.	Combine all using zip().
	2.	Convert to list.
	3.	Use zip_longest() (extra credit).
	4.	Print reversed records.
	5.	Validate using any() / all().
'''
from itertools import zip_longest
ids   = [101,102,103,104]
names = ["Amit","Riya","Gaurav"]
city  = ["kolkata","Delhi","Pune","Goa","Mumbai"]

#1.	Combine all using zip().
#2.	Convert to list.
z=list(zip(ids,names,city))

#3.	Use zip_longest() (extra credit).
p=list(zip_longest(ids,names,city,fillvalue="NA"))

#4.	Print reversed records.
print(list(reversed(p)))

#	5.	Validate using any() / all().
any_miss=any("NA" in i for i in p)
all_valid=all("NA" not in i for i in p)

print(z)
print(p)
print(any_miss)
print(all_valid)
```
