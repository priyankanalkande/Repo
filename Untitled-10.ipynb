```python
'''employees.txt
101,John,5000
102,Alice,7000
103,Bob,abc
104,Emma,8000
105,David,3000


[
    {i"d":101,"name":John,"salary":5000},
    {}
]

with open :
    for line_no,line in enumurate(f,start=1):
        line=line.strip()
        parts line

Read file line by line
Store valid employees in a list of dictionaries
Validate salary (must be integer)
Apply bonus:
salary >= 7000 → 20%
salary >= 5000 → 10%
else → 5%
Write output to salary_output.txt
Log errors and processing steps in app.log
'''

def is_num(val):
     try:
         int(val)
         return True
     except ValueError:
         return False
def bonus(val1):
    if val1>=7000:
          return val1*1.20
    elif val1>=5000:
         return val1*1.10
    else:
         return val1*1.05
             
with open("employee.txt","r") as f, open("salary_output.txt","w") as out:
         for line in f:
            emp_id,emp_name,salary=line.strip().split(",")
            #part=lines.split(",")
            #emp_id,emp_name,salary=lines
            if is_num (salary):
                 int_sal=int(salary)
                 emp_dict={
                  "id":int(emp_id),
                   "name":emp_name,
                  "sal":int_sal
                  }
                 grade=bonus(int_sal) 
                 out.write(f"{emp_dict} bonus:{grade}\n")    
```
