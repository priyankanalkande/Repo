```python
from pyspark.sql import SparkSession

# Create SparkSession in local mode with longer timeout
spark = SparkSession.builder \
    .master("local[*]") \
    .appName("RDD") \
    .config("spark.network.timeout", "600s") \
    .getOrCreate()

sc = spark.sparkContext

# Sample data
num = [10, 20, 30, 40, 50, 60]

# Create RDD with 3 partitions
rdd = sc.parallelize(num, 3)

# Actions
print("Collect:", rdd.collect())               # All elements
print("Partitions:", rdd.getNumPartitions())   # Number of partitions
print("Glom:", rdd.glom().collect())           # Elements grouped by partition
```

```python
print(rdd.glom().collect())
```
