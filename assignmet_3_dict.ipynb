```python
#Frequency Problems
# 31.	Count frequency of characters in a string.

b = {"name": "niliam", "city": "pune"}

for k, v in b.items():
    freq = {}
    for char in v:
        #print(char)
        freq[char] = freq.get(char, 0) + 1
    print(f"{k}: {freq}")
```

```python
#32.	Count frequency of words in a sentence.

b={"name":"nilam is very good good girl"}

for k,v in b.items():
    freq={}
    word=v.split()
    #print(word)
    for words in word:
        freq[words]= freq.get(words,0)+1
    print(freq)
       
    
```

```python
#33.	Find most frequent word.
b={"name":"nilam is very good good girl"}

for k,v in b.items():
    freq={}
    word=v.split()
    #print(word)
    for words in word:
        freq[words]= freq.get(words,0)+1
    for f,v in freq.items():
        if v>1:
            print(v,f)
```

```python
#34.	Remove words with frequency = 1.

b={"name":"nilam is very good good girl"}

result=[]

for k,v in b.items():
    freq={}
    word=v.split()
    #print(word)
    for words in word:
        freq[words]= freq.get(words,0)+1
    for f,v in freq.items():
        if v!=1:
            result.append((f,v))
            print(freq)
print(dict(result))
```

```python
#	35.	Sort words by frequency.

b = {"name": "nilam is very good good girl"}

for k, v in b.items():
    freq = {}
    words = v.split()
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    
    # Sort by frequency (highest first)
    sorted_words = sorted(freq.items(), key=lambda x: x[1], reverse=True)
    
    print(f"{k}: {sorted_words}")
```

```python
'''PART 3 — HARD LEVEL (61–100)

Scenario Based

Student Management System
	61.	Create student dictionary with subjects.
	62.	Calculate total marks for each student.
	63.	Calculate average marks.
	64.	Find topper of class.
	65.	Find student with lowest marks. '''

s = {
    101: {"name": "Priyanka", "maths": 85, "science": 90, "english": 75},
    102: {"name": "Nilam", "maths": 88, "science": 60, "english": 85},
    103: {"name": "Priya", "maths": 30, "science": 0, "english": 5}
}

for id, details in s.items():
    total = details["maths"] + details["science"] + details["english"]
    print(id, details["name"], "Total:", total)
    
```

```python
#63.	Calculate average marks.
for id,details in s.items():
	total=details["maths"] +details["science"]+details["english"]
    n=total/3
    print(n)
```

```python

# 64. Find topper of class
topper_name = ""
topper_id = None
topper_total = 0

for id, details in s.items():
    total = details["maths"] + details["science"] + details["english"]
    if total > topper_total:
        topper_total = total
        topper_name = details["name"]
        topper_id = id

print("Topper:", topper_name, "(ID:", topper_id, "Total Marks:", topper_total, ")")
```

```python

# 65. Find student with lowest marks
lowest_id = None
lowest_name = ""
lowest_total = float("inf")   # start with infinity so any total will be smaller

for id, details in s.items():
    total = details["maths"] + details["science"] + details["english"]
    if total < lowest_total:
        lowest_total = total
        lowest_id = id
        lowest_name = details["name"]

print("Lowest scorer:", lowest_name, "(ID:", lowest_id, "Total Marks:", lowest_total, ")")
```

```python
#69.	Remove out-of-stock product.

# Dictionary with stock included
q = {
    "Laptop": {"price": 55000, "stock": 10},
    "Mobile": {"price": 25000, "stock": 0},   # out of stock
    "Chair": {"price": 3000, "stock": 50},
    "Book": {"price": 500, "stock": 100}
}

# Remove out-of-stock products
for product, details in list(q.items()):
    print(product,details)
    if details["stock"] == 0:
        removed = q.pop(product)
        print("Removed:", product)

print("Updated dictionary:", q)
```

```python
#70.	Calculate total inventory value.
q = {
    "Laptop": {"price": 55000, "stock": 10},
    "Mobile": {"price": 25000, "stock": 0},   # out of stock
    "Chair": {"price": 3000, "stock": 50},
    "Book": {"price": 500, "stock": 100}
}

total_inventory_value = 0

for product, details in q.items():
    total_inventory_value += details["price"] * details["stock"]

print("Total inventory value:", total_inventory_value)

```
