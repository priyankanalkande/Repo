```python
'''75.	Given list of tuples (city,temp), find average temp per city (duplicates exist).'''

a = [("pune", 30), ("hyd", 20), ("zarkhand", 50), ("zarkhand", 50)]

city_sum = {}
city_count = {}

for city, temp in a:
    # add temperature to sum
    if city in city_sum:
        city_sum[city] += temp
        city_count[city] += 1
    else:
        city_sum[city] = temp
        city_count[city] = 1

# calculate average
avg_temp = {}
for city in city_sum:
    avg_temp[city] = city_sum[city] / city_count[city]

print(avg_temp)
```

```python
'''    1. Banking Transaction System
	    You receive transactions as a tuple:
        transactions = (("credit", 5000), ("debit", 2000), ("debit", 4000))

        Write logic to:
	•	Maintain balance
	•	Reject debit if insufficient balance
	•	Count rejected transactions


    Add rule:
	•	If debit > 10,000 → flag as “Fraud Alert”
	Print:
	•	Total credits
	•	Total debits
	•	Final balance
	•	Number of rejected transactions
'''

transactions = (("credit", 5000), ("debit", 2000), ("debit", 4000))

credits=0
debits=0
total=0
for i,x in transactions:
    total+=x
    if i=="debit":
        debits+=1  
        if x>10000:
              print("Fraud Alert")
    elif i=="credit":
         credits+=1

print(credits)
print(debits)
print(total)
    
```

```python
transactions = (("credit", 5000), ("debit", 2000), ("debit", 4000))

balance = 0
credits = 0
debits = 0
rejected = 0

for t_type, amount in transactions:
    if t_type == "credit":
        balance += amount
        credits += amount   # add to total credits

    elif t_type == "debit":
        if amount > 10000:
            print("Fraud Alert:", amount)
        elif amount <= balance:
            balance -= amount
            debits += amount   # add to total debits
        else:
            print("Rejected debit:", amount)
            rejected += 1

print("Total credits:", credits)
print("Total debits:", debits)
print("Final balance:", balance)
print("Number of rejected transactions:", rejected)
```

```python
'''6. Stock Market Data

    prices = (100, 105, 102, 108, 110, 107)

    Find:
	•	Maximum profit (buy low sell high)
	•	Count how many days price increased.

	-	Detect if market trend is:

	•	Increasing
	•	Decreasing
	•	Mixed '''

prices = (100, 105, 102, 108, 110, 107)

# Maximum profit (buy low, sell high)
min_price = prices[0]
max_profit = 0
for p in prices:
    if p < min_price:
        min_price = p
    profit = p - min_price
    if profit > max_profit:
        max_profit = profit

# Count how many days price increased
increase_count = 0
for i in range(1, len(prices)):
    if prices[i] > prices[i-1]:
        increase_count += 1

# Detect market trend
if all(prices[i] > prices[i-1] for i in range(1, len(prices))):
    trend = "Increasing"
elif all(prices[i] < prices[i-1] for i in range(1, len(prices))):
    trend = "Decreasing"
else:
    trend = "Mixed"

print("Maximum Profit:", max_profit)
print("Days price increased:", increase_count)
print("Market Trend:", trend)
```
