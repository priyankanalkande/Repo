```python
'''def bonus(salary):
    # salary < 30000 → "No Bonus"
    # salary 30000–50000 → salary * 0.10
    # salary > 50000 → salary * 0.20'''

def bonus(salary):
    if salary < 30000:
        return "No Bonus"
    elif 30000<salary<50000:
        return salary*0.10
    elif salary >50000:
        return salary * 0.20
    
bonus(100000)
    
```

```python
'''def signal(color):
    # "red" → "Stop"
    # "yellow" → "Ready"
    # "green" → "Go"
    # else → "Invalid Signal"'''

def signal(color):
    if color=="red":
        return "stop"
    elif color=="yellow":
        return "ready"
    elif color =="green":
        return "go"
    else:
        return "Invalid signal"
    
signal("yellow")
```

```python
'''def withdraw(balance, amount):
    # If amount > balance → "Insufficient Funds"
    # If amount <= 0 → "Invalid Amount"
    # Else → return remaining balance'''

def withdraw(balance, amount):
    if amount > balance:
        return "Insufficient Funds"
    elif amount<=0:
        return "Invalid Amount"
    else:
        return balance - amount 
    
withdraw(1000,000)
```
