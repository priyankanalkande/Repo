```python
import os

import sys

os.environ["PYSPARK_PYTHON"] = sys.executable

os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable
```

```python
from pyspark.sql import SparkSession

# Create SparkSession
spark = SparkSession.builder \
        .appName("mappractice3") \
        .getOrCreate()

sc = spark.sparkContext
```

```python
'''31. Extract Salary
Input:
[
{"name":"John","salary":50000},
{"name":"Alice","salary":70000},
{"name":"Bob","salary":60000}
]

Expected Output:
[50000,70000,60000]'''

data=[
{"name":"John","salary":50000},
{"name":"Alice","salary":70000},
{"name":"Bob","salary":60000}
]
rdd =sc.parallelize(data)
result=rdd.map(lambda x:x["salary"])
print(result.collect())
```

```python
'''Q6. Split Sentences into Words
Input:
[
"Spark is Fast",
"Python is Easy",
"AWS is Powerful"
]

Expected Output:
[
"Spark",
"is",
"Fast",
"Python",
"is",
"Easy",
"AWS",
"is",
"Powerful"
]
'''
data=[
"Spark is Fast",
"Python is Easy",
"AWS is Powerful"
]
rdd=sc.parallelize(data)
result=rdd.flatMap(lambda x:x.split(" "))
print(result.collect())
```

```python
'''Q7. Split CSV Records
Input:
[
"101,John,50000",
"102,Alice,60000",
"103,Bob,70000"
]

Expected Output:
[
"101",
"John",
"50000",
"102",
"Alice",
"60000",
"103",
"Bob",
"70000"
]'''

data=[
"101,John,50000",
"102,Alice,60000",
"103,Bob,70000"
]

rdd=sc.parallelize(data)
result=rdd.flatMap(lambda x:x.split(",")).collect()
print(result)
```

mapPartitions() Questions

```python
'''Q13. Multiply Every Number by 10
Input:
[
1,2,3,4,5,6
]

Expected Output:
[
10,20,30,40,50,60
] '''

data=[1,2,3,4,5,6]
rdd=sc.parallelize(data,2)
def multiply(num):
    for x in num:
        yield x*10

result=rdd.mapPartitions(multiply).collect()
print(result)
```

```python
'''Q14. Find Sum of Each Partition

Input:
[
10,20,30,40,50,60
]

Assume 2 Partitions

Partition 1
10
20
30

Partition 2
40
50
60

Expected Output:
[
60,
150
]'''

data=[10,20,30,40,50,60]
rdd=sc.parallelize(data,2)
def summ(num):
    yield sum(num)

result=rdd.mapPartitions(summ).collect()
print(result)
```

```python
'''Q15. Count Records in Each Partition

Input:
[
"A",
"B",
"C",
"D",
"E",
"F"
]

Assume 2 Partitions

Expected Output:
[
3,
3
]'''

data =[
"A",
"B",
"C",
"D",
"E",
"F"
]

rdd=sc.parallelize(data,2)
def countt(num):
    yield len(list(num))

result=rdd.mapPartitions(countt).collect()
print(result)
```

```python
'''Q16. Convert Every Word to Uppercase

Input:
[
"spark",
"python",
"aws",
"hadoop"
]

Expected Output:
[
"SPARK",
"PYTHON",
"AWS",
"HADOOP"
]'''

data=[
"spark",
"python",
"aws",
"hadoop"
]

rdd=sc.parallelize(data,2)
def upperr(num):
    for x in num:

         yield x.upper()

result=rdd.mapPartitions(upperr).collect()
print(result)
```

```python
'''Q17. Add 100 to Every Number

Input:
[
10,
20,
30,
40
]

Expected Output:
[
110,
120,
130,
140
] '''

data=[
10,
20,
30,
40
]

rdd=sc.parallelize(data,2)

def add(num):
    for x in num:
        yield x+100

result =rdd.mapPartitions(add).collect()
print(result)
```
