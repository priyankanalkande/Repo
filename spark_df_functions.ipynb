```python
'''dropDuplicates()

Definition  Deduplicates by all columns or specified keys; the retained non-key 
values are not deterministically chosen.

Syntax
df.dropDuplicates(["emp_id"])

Practice questions
1. Remove full duplicates from dup_df.
2. Deduplicate dup_df by id.
3. Deduplicate dup_df by name and department.
4. Deduplicate sales_df by emp_id and product.
5. Construct multiple versions per employee; observe why dropDuplicates is unsafe for latest-record selection.
Why is this unsafe?

dropDuplicates() does not guarantee which record is retained.

For employee 101, Spark may keep:

50000 salary record
60000 salary record
70000 salary record

depending on partitioning and execution plan.

So if your goal is:

"Keep the latest employee version"

then dropDuplicates() is the wrong approach because it does not consider updated_date.

6. Use a timestamp and window row_number() to select a deterministic latest record.
'''

#dup_df.dropDuplicates(["id"]).show()
#dup_df.dropDuplicates(["name","department"]).show()
#dup_df.dropDuplicates().show()
from pyspark.sql.window import Window
df.withColumn(
     "rn",
     F.row_number().over(
          Window.partitionBy("emp_id") 
           .orderBy(F.col("join_date").desc() )
           )
).filter(F.col("rn")==1).show()
```
