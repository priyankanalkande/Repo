```python
'''Q1. Multiply every number by 10.

Input:
[1, 2, 3, 4, 5]

Output:
[10, 20, 30, 40, 50] '''

from pyspark.sql import SparkSession

# Create SparkSession
spark = SparkSession.builder \
        .appName("mappractice3") \
        .getOrCreate()

sc = spark.sparkContext

# Input data
num = [1, 2, 3, 4, 5]

# Create RDD
rdd = sc.parallelize(num)

# Transformation: multiply each element by 10
result = rdd.map(lambda x: x * 10)

# Actions
print("Result:", result.collect())          # Final output
print("Partitions:", result.getNumPartitions())  # Number of partitions
print("Glom:", result.glom().collect())     # Partition-wise data
```

```python
'''Q8. Convert Celsius to Fahrenheit.

Input:
[0, 25, 100]

Output:
[32, 77, 212] '''

celsius = [0, 25, 100]

# Create RDD
rdd = sc.parallelize(celsius)

# Transformation: Convert Celsius to Fahrenheit
# Formula: F = (C * 9/5) + 32
fahrenheit_rdd = rdd.map(lambda c: (c * 9/5) + 32)

# Action: Collect results
print("Fahrenheit:", fahrenheit_rdd.collect())
```

```python
'''Q22. Add 20% tax field.

Input:
[
{"name":"John","salary":50000},
{"name":"Alice","salary":70000}
]

Output:
[
{"name":"John","salary":50000,"tax":10000},
{"name":"Alice","salary":70000,"tax":14000}
]  '''

data=[
{"name":"John","salary":50000},
{"name":"Alice","salary":70000}
]

rdd=sc.parallelize(data)
result = rdd.map(lambda x: {**x, "tax": int(x["salary"] * 0.20)})
print(result.collect())
```
